```python
# Importação das bibliotecas

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

from pathlib import Path

# Caminho para a pasta principal do projeto
PROJECT_DIR = Path.cwd().parent

# Carregamento do conjunto de treino
# Limitamos a leitura aos 188.318 registros válidos.
train = pd.read_csv(
    PROJECT_DIR / "train.csv",
    nrows=188318
)

# Carregamento do conjunto de teste
# Limitamos a leitura aos 125.546 registros válidos,
# excluindo o último registro, que está corrompido.
test = pd.read_csv(
    PROJECT_DIR / "test.csv",
    nrows=125546
)

# Conferência das dimensões
print("Dimensões do treino:", train.shape)
print("Dimensões do teste:", test.shape)
```

## Etapa 1. Conhecer a estrutura dos dados
Primeiro, vamos verificar os tipos de variáveis e entender como o Pandas está interpretando cada coluna.

```python
# Informações gerais do conjunto de treino e teste
train.info()
test.info()
```

## Etapa 2. Verificar valores ausentes
Agora vamos descobrir se existem células sem informação nos dados.

```python
# Comparação de valores ausentes entre treino e teste

missing_comparison = pd.DataFrame({
    "Treino": train.isnull().sum(),
    "Teste": test.isnull().sum()
})

# Exibir o total de valores ausentes em cada conjunto
print("Total de valores ausentes:")
print("Treino:", missing_comparison["Treino"].sum())
print("Teste:", missing_comparison["Teste"].sum())

# Exibir somente as colunas que possuem valores ausentes
missing_comparison = missing_comparison[
    (missing_comparison["Treino"] > 0) |
    (missing_comparison["Teste"] > 0)
].sort_values(by="Treino", ascending=False)

display(missing_comparison)
```

## Etapa 3. Análise das variáveis
Agora vamos identificar os tipos de variáveis presentes nos dados e entender como elas estão organizadas.

### 3.1. Identificação dos tipos de variáveis
Vamos identificar e contar as variáveis categóricas, contínuas, o identificador e a variável-alvo.

```python
# Identificar os tipos de variáveis

categorical_columns = train.filter(regex="^cat").columns
continuous_columns = train.filter(regex="^cont").columns

print("Quantidade de variáveis:")
print("Categóricas:", len(categorical_columns))
print("Contínuas:", len(continuous_columns))
print("Identificador:", "id" in train.columns)
print("Variável-alvo:", "loss" in train.columns)
```
